# What's ETL or ELT?

ETL stands for Extract, Transform, Load. In the context of Spark, ETL refers to the process of extracting data from various sources, transforming it into a desired format or structure, and loading it into a target system, such as a data warehouse or a data lake.

Here's a breakdown of each step in the ETL process:

## Extract
This step involves extracting data from multiple sources, such as databases, files (CSV, JSON, Parquet), APIs, or streaming data sources. Spark provides connectors and APIs to read data from a wide range of sources, allowing you to extract data in parallel and efficiently handle large datasets.

## Transform
In the transform step, the extracted data is processed and transformed according to specific business logic or requirements. This may involve cleaning the data, applying calculations or aggregations, performing data enrichment, filtering, joining datasets, or any other data manipulation operations. Spark provides a powerful set of transformation functions and SQL capabilities to perform these operations efficiently in a distributed and scalable manner.

## Load
Once the data has been transformed, it is loaded into a target system, such as a data warehouse, a data lake, or another storage system. Spark allows you to write the transformed data to various output formats and storage systems, including databases, distributed file systems (like Hadoop Distributed File System or Amazon S3), or columnar formats like Delta Lake or Apache Parquet. The data can be partitioned, sorted, or structured to optimize querying and analysis.

Spark's distributed computing capabilities, scalability, and rich ecosystem of libraries make it a popular choice for ETL workflows. It can handle large-scale data processing, perform complex transformations, and efficiently load data into different target systems.

By leveraging Spark for ETL, organizations can extract data from diverse sources, apply transformations to ensure data quality and consistency, and load the transformed data into a central repository for further analysis, reporting, or machine learning tasks.

# Interview Questions

While completing the rest of the practice, try to answer the following questions:

## Concepts
- What is ETL? (Hint: Explain each step)  
ETL stands for Extract, Transform, and Load.
Extract means collecting data from different sources,
Transform means cleaning and processing the data,
and Load means storing the data into a target system like a data warehouse.

## Databricks
- What is Databricks?  
Databricks is a cloud-based data platform built on top of Apache Spark. It is used for big data processing, ETL, and analytics.
- What is a Notebook?  
A notebook is an interactive environment where you can write and run code, such as Python or SQL, and visualize results.
- What is DBFS?  
DBFS is a distributed file system in Databricks used to store and access data across the cluster.
- What is a cluster?   
A cluster is a group of machines that work together to process data in parallel using Spark.
It consists of a driver node and multiple worker nodes.
In Databricks, clusters are used to run notebooks and execute data processing jobs.
- Is Databricks a data lake or a data warehouse?  
Databricks is a lakehouse platform that combines both data lake and data warehouse features.
It allows you to store raw data like a data lake and also run structured queries like a warehouse.
In practice, I used it to both store data and perform ETL and analytics in the same platform.

## Managed Table
- What is a managed table in Databricks?  
A managed table is a table where Databricks manages both the data and the metadata.
The data is stored in a default location and is fully controlled by the system.
If the table is dropped, both the metadata and the data will be deleted.
- Can you explain how to create a managed table in Databricks?  
A managed table can be created using SQL with CREATE TABLE AS SELECT, or using DataFrame write with saveAsTable.
Spark will automatically store the data and register the table in the metastore.
I used saveAsTable when building ETL pipelines in my practice.
- Can you compare a managed table with an RDBMS table? (Hint: Schema on read vs schema on write)  
In traditional RDBMS, schema is enforced before writing data, which is schema-on-write.
In big data systems like Spark, schema can be applied when reading data, which is schema-on-read.
This makes big data systems more flexible when handling unstructured or semi-structured data.
- What is the Hive metastore and how does it relate to managed tables in Databricks?  
Hive metastore is a central repository that stores metadata about tables, such as schema and table location.
It does not store the actual data, only the table definitions.
Databricks uses it to manage and access tables consistently.
- How does a managed table differ from an unmanaged (external) table in Databricks? (Hint: Consider what happens to the data when the table is deleted)  
In a managed table, Databricks controls the data, and deleting the table removes the data as well.
In an external table, only the metadata is removed when the table is dropped, but the data remains.
External tables are useful when data is stored outside Databricks.
- How can you define a schema for a managed table?  
You can either let Spark automatically infer the schema or define it manually.
With inferSchema, Spark will automatically detect the data types instead of treating everything as strings.
I usually use it for quick exploration, but in production I prefer defining the schema manually to ensure accuracy and consistency.

## Spark
`df = spark.read.format("csv").option("header", "true").option("inferSchema", "true").load(file_location)`
- What does the option("inferSchema", "true") do?  
The inferSchema option tells Spark to automatically detect the data types of each column.
Without it, all columns are treated as strings by default.
I usually enable it when exploring data to get correct data types quickly.
- What does the option("header", "true") do?  
The header option tells Spark that the first row of the file contains column names.
If it is set to false, Spark will treat the first row as actual data.
I usually set it to true when working with CSV files.
- How can you write data to a managed table?  
You can write data to a managed table using df.write.saveAsTable.
This will store the data and register the table in the metastore.
I used this approach to store processed data in my ETL workflow.
- How can you read data from a managed table into a DataFrame?  
You can read data from a managed table using spark.table.
It returns a DataFrame that you can further transform or analyze.
This is commonly used in ETL pipelines.

# Managed Table
This is an important interview topic. Some people may refer to managed tables as Hive tables.

https://docs.databricks.com/data-governance/unity-catalog/create-tables.html

```python
from pyspark.sql.types import StructType, StructField, IntegerType, TimestampType

# Define schema
schema = StructType([
    StructField("bookid", IntegerType(), True),
    StructField("facid", IntegerType(), True),
    StructField("memid", IntegerType(), True),
    StructField("starttime", TimestampType(), True),
    StructField("slots", IntegerType(), True)
])

# Read from existing table (replace CSV read)
df = spark.table("workspace.pgx.bookings")

# Apply schema (cast to match original behavior)
df = df.selectExpr(
    "cast(bookid as int) as bookid",
    "cast(facid as int) as facid",
    "cast(memid as int) as memid",
    "cast(starttime as timestamp) as starttime",
    "cast(slots as int) as slots"
)

# Drop table if exists
spark.sql("DROP TABLE IF EXISTS bookings")

# Write to managed table
df.write.saveAsTable("bookings")
```
